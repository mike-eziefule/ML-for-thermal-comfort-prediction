STEP 3 — Preprocess Data
Convert TSV (thermal sensation vote) into 3‑class labels:

Cold (−3 to −1)

Neutral (0)

Warm (+1 to +3)

```python
def classify_tsv(x):
    if x < 0:
        return "Cold"
    elif x == 0:
        return "Neutral"
    else:
        return "Warm"

df["label"] =  df["tsv"].apply(classify_tsv)
```
